# Exploratory Data Analysis (EDA)


## 1. Missing Values


```python
print('Patient Dataframe Missing Values:')
display(patient.isnull().sum())
print('\nAnnotation Dataframe Missing Values:')
display(annot.isnull().sum())
```

## 2. Dataset Overview Details


```python
num_patients = patient['patient_id'].nunique()
num_recordings = patient['recording_id'].nunique()
num_annotations = len(annot)
unique_labels = annot['clinical_label'].unique().tolist()

print(f'Total Unique Patients: {num_patients}')
print(f'Total Recordings: {num_recordings}')
print(f'Total Annotations: {num_annotations}')
print(f'Clinical Labels: {unique_labels}')
```

## 3. Patient Demographics & Recording Stats


```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gender Distribution
patient['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=axes[0], title='Gender Distribution')
axes[0].set_ylabel('')

# Age Distribution
patient['age_years'].plot(kind='hist', bins=20, ax=axes[1], title='Age Distribution', color='skyblue', edgecolor='black')
axes[1].set_xlabel('Age (Years)')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Chest Location Distribution
patient['chest_location'].value_counts().plot(kind='bar', ax=axes[0], color='lightgreen', edgecolor='black')
axes[0].set_title('Chest Location Distribution')
axes[0].set_xlabel('Chest Location')
axes[0].set_ylabel('Count')

# Overall Quality Distribution
patient['overall_quality'].value_counts().plot(kind='bar', ax=axes[1], color='salmon', edgecolor='black')
axes[1].set_title('Overall Quality Distribution')
axes[1].set_xlabel('Quality')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
patient.groupby(['dataset_split', 'gender']).size().unstack().plot(kind='bar', stacked=False, figsize=(10, 5))
plt.title('Train/Test Split by Gender')
plt.xlabel('Dataset Split')
plt.ylabel('Count')
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

## 4. Annotation Details


```python
plt.figure(figsize=(10, 5))
annot['clinical_label'].value_counts().plot(kind='barh', color='purple', edgecolor='black')
plt.title('Distribution of Clinical Labels')
plt.xlabel('Count')
plt.ylabel('Clinical Label')
plt.gca().invert_yaxis()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
annot.boxplot(column='duration_ms', by='clinical_label', grid=False, rot=45, figsize=(12, 6))
plt.title('Duration of Annotations by Clinical Label')
plt.suptitle('')
plt.xlabel('Clinical Label')
plt.ylabel('Duration (ms)')
plt.show()
```
