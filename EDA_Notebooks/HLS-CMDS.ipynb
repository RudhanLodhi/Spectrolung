# Visualize sounds

```python
import os
import random
from IPython.display import Audio, display

def display_random_sounds(df, id_column, sound_dir, category_name, n=3):
    print(f"--- 3 Random {category_name} Sounds ---")
    
    # Sample n random rows from the dataframe
    sampled_df = df.sample(n=min(n, len(df)))
    
    for _, row in sampled_df.iterrows():
        sound_id = row[id_column]
        file_path = os.path.join(sound_dir, f"{sound_id}.wav")
        
        # Display the labels associated with this sound
        labels = []
        if 'Heart Sound Type' in row:
            labels.append(f"Heart: {row['Heart Sound Type']}")
        if 'Lung Sound Type' in row:
            labels.append(f"Lung: {row['Lung Sound Type']}")
            
        print(f"ID: {sound_id} | {' | '.join(labels)}")
        
        # Display the interactive audio player
        if os.path.exists(file_path):
            display(Audio(file_path))
        else:
            print(f"File not found: {file_path}")
        print("-" * 40)
    print("\n")

# Display 3 random Mixed Sounds
display_random_sounds(mix, 'Mixed Sound ID', Mix_sounds, 'Mixed', n=3)

# Display 3 random Heart Sounds
display_random_sounds(hs, 'Heart Sound ID', heart_sounds, 'Heart', n=3)

# Display 3 random Lung Sounds
display_random_sounds(ls, 'Lung Sound ID', lung_sounds, 'Lung', n=3)
```

```python
# Example: 3 random sounds for each Lung Sound Type
print("--- 3 Random Sounds per Lung Sound Type ---")
for sound_type in ls['Lung Sound Type'].unique():
    print(f"\n### Category: {sound_type} ###")
    
    # Filter the dataset for the current sound type and sample up to 3
    type_df = ls[ls['Lung Sound Type'] == sound_type]
    sampled_df = type_df.sample(n=min(3, len(type_df)))
    
    for _, row in sampled_df.iterrows():
        sound_id = row['Lung Sound ID']
        file_path = os.path.join(lung_sounds, f"{sound_id}.wav")
        print(f"ID: {sound_id}")
        if os.path.exists(file_path):
            display(Audio(file_path))
```
